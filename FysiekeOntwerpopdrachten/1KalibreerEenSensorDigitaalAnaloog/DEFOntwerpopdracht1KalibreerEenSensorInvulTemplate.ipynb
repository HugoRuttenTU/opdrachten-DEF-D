## *Opdracht 6*: Meten en fitten, Leerdoel

```python
# Zet hier jouw data in een voor jou handig format
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
lichtsterkte = np.array([7.5,7.5,7.0,7.7,7.9,7.7,8.7,8.5,8.6,6.9,7.0,7.0,7.2,7.1,7.1,7.3,7.3,7.3,7.4,7.6,7.5,5.9,5.8,5.5,3.3,3.3,3.1,4.6,4.0,3.9,4.0,5.4,5.2,5.1,5.3,6.3,6.3,6.4,2.3,3.0,3.1,8.1,7.9,7.9,8.0,8.5,8.4,8.4 ]) #EV
lux = 2.5*2**lichtsterkte
bits = np.array([933,938,925,957,962,957,976,975,976,926,929,926,933,929,932,940,947,948,946,948,949,871,857,844,609,609,605,759,737,721,725,825,814,811,829,887,893,891,557,586,597,958,948,951,954,964,965,962   ])

```

```python
#onzekerheden
u_EV = 0.2
u_bits = 19

#curve fit
def fit(x,a,b):
        return a*np.log(x)+b

popt, pcov = curve_fit(fit,lux,bits)
xfit = np.linspace(min(lux), max(lux),1000)
yfit = fit(xfit,*popt)

#plot
rel_error_high = 2**u_EV -1
rel_error_low = -1/2**u_EV + 1

lux_error_low = rel_error_low * lux
lux_error_high = rel_error_high * lux

plt.errorbar(lux, bits,xerr=[lux_error_low,lux_error_high],yerr = 19,fmt='.',label='data')
plt.plot(xfit,yfit,'r--',label='curvefit')
plt.xlabel('lichtsterkte (lx)')
plt.ylabel('bits')
#plt.xscale('log')
plt.legend()
plt.grid()
plt.show

print(*popt)
```

## *Opdracht 8*: Conclusie, leerdoel
